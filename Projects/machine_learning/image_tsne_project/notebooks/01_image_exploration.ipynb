```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# I add the project root to the system path to locate the data folder
project_root = os.path.abspath('..')
if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
# I load the raw, flattened pixel data that I generated in data.py
raw_data_path = os.path.join(project_root, "data", "raw", "fashion_mnist_raw.csv")
df = pd.read_csv(raw_data_path)

print(f"Dataset shape: {df.shape}")
print("I expect 10,000 rows and 786 columns (784 pixels + label + category).")
```

```python
# I isolate the pixel columns
pixel_cols = [col for col in df.columns if col.startswith('pixel')]
pixels = df[pixel_cols].values

# I extract the text labels to use as titles for the subplots
categories = df['category'].values

# I set up a 2x5 grid using Matplotlib to display 10 sample images
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

# I loop through the first 10 rows of my dataset
for i, ax in enumerate(axes.flat):
    # I take the flat 784-item array and reshape it back into a 28x28 2D array
    image_matrix = pixels[i].reshape(28, 28)

    # I plot the image using a grayscale color map
    ax.imshow(image_matrix, cmap='gray')
    ax.set_title(categories[i])
    ax.axis('off')

plt.tight_layout()
plt.show()
```
