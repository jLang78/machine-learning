```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

# Load API key securely
project_root = os.path.abspath('..')
load_dotenv(os.path.join(project_root, '.env'))
NEWS_API_KEY = os.getenv("NEWS_API_KEY")

print("API Key Loaded Successfully" if NEWS_API_KEY else "Missing API Key!")
```

```python
# Test query for a single topic
query = "technology"
url = f"https://newsapi.org/v2/everything?q={query}&language=en&pageSize=5&apiKey={NEWS_API_KEY}"

response = requests.get(url)
data = response.json()

print(f"Status: {data.get('status')}")
print(f"Total Results Found: {data.get('totalResults')}")
```

```python
# Looking at the first article's keys
if data.get('status') == 'ok' and data.get('articles'):
    first_article = data['articles'][0]
    print("Keys available in each article:")
    for key, value in first_article.items():
        print(f" - {key}: {type(value).__name__}")
```

```python
# Extracting just the title and description
records = []
for article in data.get('articles', []):
    records.append({
        "topic": query,
        "title": article.get('title'),
        "description": article.get('description')
    })

df_prototype = pd.DataFrame(records)
df_prototype.head()
```
